## 0. Cadre éthique et réglementaire (rappel du cours)
Les points du visage sont des **données biométriques**, donc des données personnelles sensibles (loi 09-08 au Maroc ; la CNDP encadre leur traitement). Afin de se conformer à la législation en vigueur, il faut veiller aux points suivants :
- **Images générées par l'IA** : dans le cadre du respect des lois en vigueur (loi 09-08 relative à la protection des personnes physiques à l'égard du traitement des données à caractère personnel, et délibérations de la CNDP), **toutes les images de `data/faces/` ont été générées par intelligence artificielle**. Elles ne représentent aucune personne réelle identifiée ou identifiable. Aucune donnée biométrique réelle n'est donc collectée ni traitée, ce qui évite les questions de consentement, d'information des personnes et de déclaration ou d'autorisation préalable auprès de la CNDP.
- **Finalité** : strictement pédagogique. Le pipeline est conçu pour pouvoir traiter de vraies données plus tard, mais dans ce cas les obligations ci-dessous et celles de la section 9 s'appliqueraient pleinement (consentement explicite, autorisation CNDP, sécurité du stockage).
- **Limite de ce choix** : un visage synthétique peut présenter des régularités, comme une symétrie ou une texture lisse, qui le distinguent d'une vraie photo. Les mesures de qualité et de distance obtenues ici ne valent donc pas comme estimation de performance sur une population réelle.
- **Minimisation** : on exporte des descripteurs géométriques, pas les images. Aucune donnée d'identité civile ne figure dans le CSV : les sujets sont désignés par des pseudonymes (`personne_1`, `personne_2`, …).
- **Traitement local** : aucun envoi vers un service tiers, et les modèles s'exécutent hors ligne. Cela suit l'esprit de la délibération CNDP n° 478-2013 (limitation du stockage centralisé) et de NIST SP 800-63B-4 (comparaison locale de préférence).
- **Non-révocabilité** : un visage ne peut pas être « changé ». Le CSV produit doit donc être protégé (voir section 9).

## 1. Environnement et configuration
**Choix des bibliothèques**
- **OpenCV version 4.8 ou supérieure** (`opencv-contrib-python`) fournit `cv2.FaceDetectorYN`, c'est-à-dire le détecteur **YuNet**, un CNN léger qui renvoie une boîte, 5 points et un score de confiance. Il fournit aussi `cv2.face.createFacemarkLBF`, la régression de **68 landmarks** par *Local Binary Features*.
- **pandas / numpy** pour les tableaux, **matplotlib** pour l'observation visuelle.

**Pourquoi pas MediaPipe ?** MediaPipe Face Mesh (468 points) est une alternative valable. Les versions récentes de la bibliothèque ne contiennent plus le modèle, qu'il faut télécharger à part. Le schéma iBUG à 68 points est aussi le plus documenté, ce qui facilite l'interprétation des descripteurs (indices des yeux, du nez, de la bouche).

**Intégrité des modèles (confiance numérique)** : chaque modèle est vérifié par son empreinte **SHA-256** avant usage. Un modèle altéré, qu'il s'agisse d'une corruption ou d'une substitution malveillante, est refusé.

## Récupération et vérification d'intégrité des modèles

```python
def sha256_file(path: Path, chunk: int = 1 << 20) -> str:
    # Empreinte SHA-256 calculée par blocs pour ne pas charger 56 Mo en mémoire d'un coup
    h = hashlib.sha256()
    with open(path, "rb") as f:
        for block in iter(lambda: f.read(chunk), b""):
            h.update(block)
    return h.hexdigest()

def ensure_model(name: str, spec: dict) -> Path:
    path = spec["file"]
    if not path.exists():
        # Téléchargement uniquement si le modèle est absent du zip (il y est normalement fourni)
        print(f"[{name}] téléchargement depuis {spec['url']}")
        urllib.request.urlretrieve(spec["url"], path)
    digest = sha256_file(path)
    if digest != spec["sha256"]:
        # Refus explicite : on n'utilise jamais un modèle dont l'intégrité n'est pas prouvée
        raise RuntimeError(f"[{name}] empreinte SHA-256 inattendue : {digest}")
    print(f"[{name}] OK  {path.name}  sha256={digest[:16]}…")
    return path

YUNET_PATH = ensure_model("yunet", MODELS["yunet"])
LBF_PATH   = ensure_model("lbf",   MODELS["lbf"])

# Chargement unique des modèles (coûteux, surtout LBF : environ 56 Mo)
detector = cv2.FaceDetectorYN.create(str(YUNET_PATH), "", (320, 320),
                                     score_threshold=0.5, nms_threshold=NMS_THRESHOLD, top_k=50)
facemark = cv2.face.createFacemarkLBF()
facemark.loadModel(str(LBF_PATH))
print("Modèles chargés.")
```

## 2. Inventaire des données (parcours automatique des dossiers)
- Le parcours de `data/faces/` est **récursif** (`os.walk`), donc tous les sous-dossiers sont traités.
- **Convention** : le premier sous-dossier sous `data/faces/` porte l'identifiant (pseudonyme) du sujet, (exemple `data/faces/personne_01/img1.jpg`). Une image posée directement à la racine reçoit le sujet `inconnu`.
- Les fichiers qui ne sont pas des images (selon l'extension) sont inventoriés et ignorés, avec le statut `SKIPPED_NOT_IMAGE`, pour que chaque exclusion reste traçable.
- Une empreinte SHA-256 du fichier identifie l'échantillon sans ambiguïté et permet de repérer les **doublons** (une même image copiée deux fois biaiserait l'évaluation).

```python
def list_files(root: Path) -> pd.DataFrame:
    rows = []
    if not root.exists():
        raise FileNotFoundError(f"Dossier introuvable : {root.resolve()}")
    for dirpath, dirnames, filenames in os.walk(root):
        dirnames.sort()                      # ordre déterministe : exécutions reproductibles
        for fn in sorted(filenames):
            if fn.startswith("."):           # fichiers cachés (.DS_Store, ...) ignorés
                continue
            p = Path(dirpath) / fn
            rel = p.relative_to(root)
            subject = rel.parts[0] if len(rel.parts) > 1 else "inconnu"
            rows.append({
                "file_relpath": rel.as_posix(),
                "subject_id": subject,
                "is_image_ext": p.suffix.lower() in IMAGE_EXTS,
                "file_size_bytes": p.stat().st_size,
            })
    return pd.DataFrame(rows)

inventory = list_files(DATA_DIR)
print(f"{len(inventory)} fichiers trouvés, dont {inventory.is_image_ext.sum()} avec une extension image")
skipped = inventory.loc[~inventory.is_image_ext, "file_relpath"].tolist()
if skipped:
    print("Fichiers ignorés (extension non image) :", skipped)
if (inventory.subject_id == "inconnu").any():
    print("Attention : des images sont à la racine de data/faces -> sujet 'inconnu'. "
          "Rangez-les dans un sous-dossier par personne (personne_01/, personne_02/, ...).")
display(inventory.groupby("subject_id").agg(n_fichiers=("file_relpath", "count"),
                                            n_images=("is_image_ext", "sum")))
```

## 3. Fonctions du pipeline
### 3.1 Chargement et prétraitement
- `np.fromfile` + `cv2.imdecode` au lieu de `cv2.imread` : cette combinaison gère les chemins contenant des accents ou des caractères non ASCII. Un décodage qui échoue renvoie `None`, ce qui produit le statut `ERROR_DECODE` au lieu de faire planter le lot.
- Les images en niveaux de gris ou avec canal alpha sont converties en BGR à 3 canaux.
- **Redimensionnement borné** (`MAX_SIDE`) : les coordonnées sont calculées sur l'image réduite. C'est sans effet sur les descripteurs, qui sont **normalisés en échelle**.
- **CLAHE** sur la luminance (espace LAB) atténue les variations d'éclairage, l'un des facteurs environnementaux cités dans le cours.

```python
def load_image(path: Path):
    # Lecture robuste. Renvoie (image BGR uint8, message d'erreur ou None).
    try:
        buf = np.fromfile(str(path), dtype=np.uint8)
        img = cv2.imdecode(buf, cv2.IMREAD_UNCHANGED)
    except Exception as e:                          # fichier illisible, droits, etc.
        return None, f"lecture impossible : {e}"
    if img is None or img.size == 0:
        return None, "décodage impossible (fichier corrompu ou format non supporté)"
    if img.dtype != np.uint8:                       # ex. PNG/TIFF 16 bits : ramené sur 8 bits
        img = cv2.normalize(img, None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)
    if img.ndim == 2:
        img = cv2.cvtColor(img, cv2.COLOR_GRAY2BGR)
    elif img.shape[2] == 4:
        img = cv2.cvtColor(img, cv2.COLOR_BGRA2BGR)
    return img, None

def preprocess(img):
    # Redimensionnement borné et version CLAHE pour les landmarks. Renvoie (img, img_clahe, facteur).
    h, w = img.shape[:2]
    scale = min(1.0, MAX_SIDE / max(h, w))
    if scale < 1.0:
        img = cv2.resize(img, (round(w * scale), round(h * scale)), interpolation=cv2.INTER_AREA)
    if USE_CLAHE:
        lab = cv2.cvtColor(img, cv2.COLOR_BGR2LAB)
        lab[..., 0] = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8)).apply(lab[..., 0])
        img_eq = cv2.cvtColor(lab, cv2.COLOR_LAB2BGR)
    else:
        img_eq = img
    return img, img_eq, scale
```

### 3.2 Détection du visage et extraction des 68 landmarks
- **YuNet** renvoie pour chaque visage `[x, y, w, h, 5 points (yeux, nez, coins de la bouche), score]`.
- **Politique en cas de visages multiples** : on garde le **plus grand** visage, supposé être le sujet enrôlé, et on enregistre `n_faces_detected` pour la traçabilité. Pour un enrôlement, une image à plusieurs visages est en soi un défaut de qualité.
- **Facemark LBF** ajuste les 68 points iBUG 300-W à partir de la boîte du détecteur.
- **Redressement du roulis** : LBF est entraîné sur des visages droits. Si la ligne des yeux est inclinée de plus de 3°, l'image est tournée pour remettre le visage droit, puis les landmarks sont ré-ajustés et ramenés dans le repère d'origine (2 à 3 itérations au maximum).
- **Contrôle de cohérence** : on compare les yeux trouvés par YuNet aux centres des yeux calculés à partir des landmarks LBF. Un grand écart signale un ajustement raté, que LBF ne signale pas de lui-même car il renvoie toujours 68 points.

Indices iBUG (0 à 67) : mâchoire 0–16, sourcils 17–26, arête du nez 27–30, base du nez 31–35, œil droit du sujet 36–41, œil gauche 42–47, lèvres extérieures 48–59, lèvres intérieures 60–67.

```python
# Régions iBUG-68 (indices 0-based)
IDX = {
    "jaw": range(0, 17), "brow_r": range(17, 22), "brow_l": range(22, 27),
    "nose_bridge": range(27, 31), "nose_base": range(31, 36),
    "eye_r": range(36, 42), "eye_l": range(42, 48),
    "lip_outer": range(48, 60), "lip_inner": range(60, 68),
}
# Permutation miroir : point i  <->  son symétrique (sert à l'indice de symétrie)
MIRROR = np.arange(68)
for a, b in [*[(i, 16 - i) for i in range(8)],
             (17, 26), (18, 25), (19, 24), (20, 23), (21, 22), (31, 35), (32, 34),
             (36, 45), (37, 44), (38, 43), (39, 42), (40, 47), (41, 46),
             (48, 54), (49, 53), (50, 52), (55, 59), (56, 58), (60, 64), (61, 63), (65, 67)]:
    MIRROR[a], MIRROR[b] = b, a

def detect_faces(img):
    h, w = img.shape[:2]
    detector.setInputSize((w, h))
    _, faces = detector.detect(img)
    return np.zeros((0, 15), np.float32) if faces is None else faces

def _fit_lbf(img_eq, box):
    # Ajustement LBF brut dans une boîte [x, y, w, h]. Renvoie (68, 2) ou None.
    gray = cv2.cvtColor(img_eq, cv2.COLOR_BGR2GRAY)
    ok, lms = facemark.fit(gray, np.array([box], dtype=np.int32))
    if not ok or len(lms) == 0:
        return None
    pts = np.asarray(lms[0], dtype=np.float64).reshape(-1, 2)   # (1,68,2) -> (68,2)
    return pts if pts.shape == (68, 2) and np.isfinite(pts).all() else None

def _eye_angle(pts):
    v = pts[list(IDX["eye_l"])].mean(0) - pts[list(IDX["eye_r"])].mean(0)
    return float(np.degrees(np.arctan2(v[1], v[0])))

def fit_landmarks(img_eq, face_row, max_iter=3, tol_deg=3.0):
    # Ajuste les 68 landmarks avec REDRESSEMENT ITÉRATIF du roulis.
    # LBF et YuNet sont entraînés sur des visages droits : au-delà d'environ 15° d'inclinaison,
    # les points dérivent (constaté sur personne_1/image_3, roll ≈ -27°). On fait donc :
    #   1) un premier ajustement, puis l'estimation de l'angle de la ligne des yeux ;
    #   2) une rotation de l'image pour remettre le visage droit, une nouvelle détection et un nouvel ajustement ;
    #   3) la transformation inverse pour revenir aux coordonnées de l'image d'origine.
    # Renvoie (pts (68,2) dans l'image d'origine, yeux YuNet (2,2) dans ce même repère) ou (None, None).
    pts = _fit_lbf(img_eq, face_row[:4])
    if pts is None:
        return None, None
    eyes = np.array([face_row[4:6], face_row[6:8]], dtype=np.float64)
    h, w = img_eq.shape[:2]
    A_total = np.eye(3)                                    # transformation cumulée : origine -> repère redressé
    img_r, face_r = img_eq, face_row
    pts_frame = pts                                        # landmarks dans le repère courant (redressé)
    for _ in range(max_iter):
        ang = _eye_angle(pts_frame)                        # angle RÉSIDUEL dans le repère courant
        if abs(ang) < tol_deg:
            break
        c = pts.mean(0)
        R = np.vstack([cv2.getRotationMatrix2D((float(c[0]), float(c[1])), ang, 1.0), [0, 0, 1]])
        A_total = R @ A_total
        img_r = cv2.warpAffine(img_eq, A_total[:2], (w, h), borderMode=cv2.BORDER_REPLICATE)
        f = detect_faces(img_r)
        if len(f):                                         # nouvelle boîte et nouveaux yeux sur le visage redressé
            face_r = f[np.argmax(f[:, 2] * f[:, 3])]
        q = _fit_lbf(img_r, face_r[:4])
        if q is None:
            break
        pts_frame = q
        A_inv = np.linalg.inv(A_total)
        to_orig = lambda P: (np.hstack([P, np.ones((len(P), 1))]) @ A_inv.T)[:, :2]
        pts = to_orig(q)
        eyes = to_orig(np.array([face_r[4:6], face_r[6:8]], dtype=np.float64))
    return pts, eyes
```

### 3.3 Descripteurs normalisés (le « gabarit » géométrique)
Les coordonnées brutes en pixels dépendent de la position du visage dans l'image, de sa taille et de l'inclinaison de la tête. Elles ne sont donc **pas comparables d'une image à l'autre**. On applique une **normalisation par similitude** (alignement de Procuste simplifié) :
1. **Translation** : l'origine est placée au milieu des deux centres des yeux.
2. **Rotation** : la droite qui relie les yeux devient horizontale. L'angle annulé est conservé dans `roll_deg`, qui sert d'indicateur de qualité.
3. **Échelle** : tout est divisé par la **distance inter-oculaire (IOD)**, qui vaut alors 1.

On obtient **136 coordonnées** `lmk_00_x … lmk_67_y` sans unité, auxquelles s'ajoutent des **ratios anthropométriques** interprétables, eux aussi rapportés à l'IOD :

| Descripteur | Définition |
|---|---|
| `face_width` | distance 0–16 (largeur de la mâchoire) |
| `face_height` | distance du milieu des sourcils au menton (8) |
| `nose_length`, `nose_width` | 27–33 et 31–35 |
| `mouth_width`, `mouth_open_ratio` | 48–54 ; ouverture intérieure 62–66 / largeur |
| `eye_r_ear`, `eye_l_ear` | *Eye Aspect Ratio* (Soukupová & Čech, 2016) |
| `brow_eye_dist` | hauteur moyenne des sourcils au-dessus des yeux |
| `jaw_face_ratio`, `face_aspect` | largeur au niveau de la bouche / largeur totale ; hauteur / largeur |
| `symmetry_index` | écart moyen entre la forme et son image miroir (0 = parfaitement symétrique) |
| `yaw_index` | asymétrie nez/bords de la mâchoire : indicateur de rotation gauche-droite |

> **Limite** : cette normalisation corrige le roulis (rotation dans le plan) mais **pas** le lacet ni le tangage (rotations hors plan). C'est pourquoi `yaw_index` est mesuré pour le contrôle qualité.

```python
def _d(a, b):
    return float(np.linalg.norm(np.asarray(a) - np.asarray(b)))

def normalize_landmarks(pts):
    # Normalisation par similitude. Renvoie (pts_norm (68,2), iod_px, roll_deg).
    eye_r = pts[list(IDX["eye_r"])].mean(axis=0)   # œil droit du sujet (à gauche dans l'image)
    eye_l = pts[list(IDX["eye_l"])].mean(axis=0)
    center = (eye_r + eye_l) / 2.0
    v = eye_l - eye_r
    iod = float(np.linalg.norm(v))
    if iod < 1e-6:
        raise ValueError("distance inter-oculaire nulle : landmarks dégénérés")
    roll = np.arctan2(v[1], v[0])                  # angle de la ligne des yeux
    c, s = np.cos(-roll), np.sin(-roll)
    R = np.array([[c, -s], [s, c]])
    pts_n = (pts - center) @ R.T / iod             # translation, rotation, puis mise à l'échelle
    return pts_n, iod, float(np.degrees(roll))

def eye_aspect_ratio(e):
    # EAR = (|p2-p6| + |p3-p5|) / (2 |p1-p4|), avec e = 6 points d'un œil
    return (_d(e[1], e[5]) + _d(e[2], e[4])) / (2.0 * _d(e[0], e[3]))

def geometric_features(p):
    # Ratios anthropométriques calculés sur les landmarks normalisés (IOD = 1)
    brow_mid = p[list(IDX["brow_r"]) + list(IDX["brow_l"])].mean(axis=0)
    face_w = _d(p[0], p[16])
    face_h = _d(brow_mid, p[8])
    brows_y = p[list(IDX["brow_r"]) + list(IDX["brow_l"])][:, 1].mean()
    eyes_y = p[list(IDX["eye_r"]) + list(IDX["eye_l"])][:, 1].mean()
    mouth_w = _d(p[48], p[54])
    dl, dr = _d(p[30], p[0]), _d(p[30], p[16])
    mirrored = p[MIRROR] * np.array([-1.0, 1.0])            # miroir vertical autour de l'axe x = 0
    return {
        "face_width": face_w,
        "face_height": face_h,
        "face_aspect": face_h / face_w,
        "nose_length": _d(p[27], p[33]),
        "nose_width": _d(p[31], p[35]),
        "mouth_width": mouth_w,
        "mouth_open_ratio": _d(p[62], p[66]) / mouth_w,
        "eye_r_ear": eye_aspect_ratio(p[list(IDX["eye_r"])]),
        "eye_l_ear": eye_aspect_ratio(p[list(IDX["eye_l"])]),
        "brow_eye_dist": float(eyes_y - brows_y),           # y croissant vers le bas
        "jaw_face_ratio": _d(p[4], p[12]) / face_w,
        "symmetry_index": float(np.linalg.norm(p - mirrored, axis=1).mean()),
        "yaw_index": (dl - dr) / (dl + dr),
    }
```

### 3.4 Indicateurs de qualité
La performance dépend de la qualité de l'échantillon : éclairage, posture, netteté. On mesure, **sur la région du visage** et non sur toute l'image :
- `det_score` : confiance du détecteur ;
- `iod_px` : résolution utile (distance inter-oculaire en pixels, avant normalisation) ;
- `sharpness` : variance du laplacien sur le visage ramené à 128×128 pour que la mesure ne dépende pas de la taille ;
- `brightness`, `contrast` : moyenne et écart-type des niveaux de gris ;
- `roll_deg`, `yaw_index` : pose ;
- `lmk_consistency` : écart YuNet/LBF sur les yeux, en fraction de l'IOD.

`quality_ok` combine ces critères, et `quality_issues` dit **lesquels** échouent. C'est un drapeau d'aide à la décision : l'échantillon n'est pas supprimé.

```python
def face_quality(img, face_row):
    h, w = img.shape[:2]
    x, y, bw, bh = [int(round(v)) for v in face_row[:4]]
    x0, y0, x1, y1 = max(0, x), max(0, y), min(w, x + bw), min(h, y + bh)
    roi = cv2.cvtColor(img[y0:y1, x0:x1], cv2.COLOR_BGR2GRAY)
    if roi.size == 0:
        return {"sharpness": np.nan, "brightness": np.nan, "contrast": np.nan}
    roi128 = cv2.resize(roi, (128, 128), interpolation=cv2.INTER_AREA)
    return {
        "sharpness": float(cv2.Laplacian(roi128, cv2.CV_64F).var()),
        "brightness": float(roi.mean()),
        "contrast": float(roi.std()),
    }

def quality_issues(r):
    issues = []
    if r["iod_px"] < Q_IOD_MIN_PX:                     issues.append("low_resolution")
    if r["sharpness"] < Q_SHARPNESS_MIN:               issues.append("blur")
    if not (Q_BRIGHT_RANGE[0] <= r["brightness"] <= Q_BRIGHT_RANGE[1]): issues.append("exposure")
    if r["contrast"] < Q_CONTRAST_MIN:                 issues.append("low_contrast")
    if abs(r["roll_deg"]) > Q_ROLL_MAX_DEG:            issues.append("roll")
    if abs(r["yaw_index"]) > Q_YAW_MAX:                issues.append("yaw")
    if r["lmk_consistency"] > Q_CONSIST_MAX:           issues.append("landmark_fit")
    if r["n_faces_detected"] > 1:                      issues.append("multiple_faces")
    return issues
```

### 3.5 Traitement d'une image (avec gestion d'erreurs)
Chaque image produit **exactement une ligne**, même en cas d'échec, avec un `status` normalisé :

| status | Signification |
|---|---|
| `OK` | landmarks extraits, descripteurs calculés |
| `FTA_NO_FACE` | aucun visage détecté au-dessus de `DET_SCORE_MIN` (*Failure To Acquire*, ISO/IEC 19795-1) |
| `FTA_LANDMARKS` | visage détecté mais ajustement des landmarks impossible ou dégénéré |
| `ERROR_DECODE` | fichier illisible ou corrompu |
| `SKIPPED_NOT_IMAGE` | extension non reconnue comme image |
| `ERROR_UNEXPECTED` | exception imprévue (message dans `error_msg`) : le lot continue |

Cette traçabilité, une ligne par fichier, permet de calculer le **taux d'échec d'acquisition (FTA)**, et répond à l'exigence de **responsabilité (*accountability*)**.

```python
def process_image(root: Path, rel: str, subject: str) -> dict:
    path = root / rel
    t0 = time.perf_counter()
    rec = {"file_relpath": rel, "subject_id": subject, "status": None, "error_msg": ""}
    try:
        rec["image_sha256"] = sha256_file(path)
        img, err = load_image(path)
        if img is None:
            rec.update(status="ERROR_DECODE", error_msg=err)
            return rec
        rec["img_height"], rec["img_width"] = img.shape[:2]
        img, img_eq, scale = preprocess(img)
        rec["resize_factor"] = scale

        faces = detect_faces(img)
        faces = faces[faces[:, 14] >= DET_SCORE_MIN] if len(faces) else faces
        rec["n_faces_detected"] = int(len(faces))
        if len(faces) == 0:
            rec.update(status="FTA_NO_FACE", error_msg="aucun visage au-dessus du seuil de détection")
            return rec
        face = faces[np.argmax(faces[:, 2] * faces[:, 3])]      # plus grand visage
        rec["det_score"] = float(face[14])
        rec["bbox_x"], rec["bbox_y"], rec["bbox_w"], rec["bbox_h"] = [float(v) for v in face[:4]]

        pts, yu_eyes = fit_landmarks(img_eq, face)
        if pts is None:
            rec.update(status="FTA_LANDMARKS", error_msg="ajustement LBF impossible")
            return rec

        pts_n, iod, roll = normalize_landmarks(pts)
        rec["iod_px"], rec["roll_deg"] = iod, roll

        # Cohérence YuNet / LBF : YuNet donne d'abord l'œil droit puis l'œil gauche (du point de vue du sujet)
        yu_r, yu_l = yu_eyes            # yeux YuNet détectés sur le visage redressé, ramenés dans l'image d'origine
        lb_r, lb_l = pts[list(IDX["eye_r"])].mean(0), pts[list(IDX["eye_l"])].mean(0)
        rec["lmk_consistency"] = (_d(yu_r, lb_r) + _d(yu_l, lb_l)) / (2 * iod)

        rec.update(face_quality(img, face))
        rec.update(geometric_features(pts_n))
        for i, (x, y) in enumerate(pts_n):
            rec[f"lmk_{i:02d}_x"], rec[f"lmk_{i:02d}_y"] = float(x), float(y)

        issues = quality_issues(rec)
        rec["quality_ok"] = len(issues) == 0
        rec["quality_issues"] = ";".join(issues) if issues else "none"
        rec["status"] = "OK"
        rec["_landmarks_px"] = pts                 # gardé en mémoire pour la visualisation, non exporté
    except Exception as e:                         # le lot ne doit jamais s'arrêter sur un seul fichier
        rec.update(status="ERROR_UNEXPECTED", error_msg=f"{type(e).__name__}: {e}")
    finally:
        rec["proc_time_ms"] = round((time.perf_counter() - t0) * 1000, 1)
    return rec
```

## 4. Observation : landmarks sur quelques images
Avant de lancer le lot, on **observe** le résultat sur un échantillon (premier objectif du TP). En rouge : les 68 landmarks LBF ; en bleu : les 5 points YuNet ; en vert : la boîte de détection.

```python
def draw_result(root, rec):
    img, _ = load_image(root / rec["file_relpath"])
    img, _, _ = preprocess(img)
    vis = img.copy()
    x, y, w, h = [int(rec[k]) for k in ("bbox_x", "bbox_y", "bbox_w", "bbox_h")]
    t = max(1, round(max(img.shape[:2]) / 400))
    cv2.rectangle(vis, (x, y), (x + w, y + h), (0, 200, 0), t)
    pts = rec["_landmarks_px"]
    # Contours reliés pour mieux lire la forme
    for name, closed in [("jaw", False), ("brow_r", False), ("brow_l", False), ("nose_bridge", False),
                         ("nose_base", False), ("eye_r", True), ("eye_l", True),
                         ("lip_outer", True), ("lip_inner", True)]:
        poly = pts[list(IDX[name])].astype(np.int32)
        cv2.polylines(vis, [poly], closed, (0, 0, 255), t)
    for p in pts:
        cv2.circle(vis, tuple(int(v) for v in p), t + 1, (0, 0, 255), -1)
    face = detect_faces(img)
    if len(face):
        f = face[np.argmax(face[:, 2] * face[:, 3])]
        for k in range(5):
            cv2.circle(vis, (int(f[4 + 2 * k]), int(f[5 + 2 * k])), t + 2, (255, 0, 0), -1)
    pad = int(0.35 * max(w, h))                    # recadrage autour du visage pour la lisibilité
    crop = vis[max(0, y - pad): y + h + pad, max(0, x - pad): x + w + pad]
    return cv2.cvtColor(crop, cv2.COLOR_BGR2RGB)

images = inventory[inventory.is_image_ext]
preview = []
for _, r in images.iterrows():
    rec = process_image(DATA_DIR, r.file_relpath, r.subject_id)
    if rec["status"] == "OK":
        preview.append(rec)
    if len(preview) == 4:
        break

if preview:
    fig, axes = plt.subplots(1, len(preview), figsize=(4 * len(preview), 4.6))
    for ax, rec in zip(np.atleast_1d(axes), preview):
        ax.imshow(draw_result(DATA_DIR, rec)); ax.axis("off")
        ax.set_title(f"{rec['file_relpath']}\nscore={rec['det_score']:.2f}  IOD={rec['iod_px']:.0f}px", fontsize=9)
    plt.tight_layout(); plt.show()
else:
    print("Aucun visage exploitable trouvé pour l'aperçu.")
```

**Points à observer** (grille de lecture) :
- Les points des **yeux, du nez et de la bouche** sont les plus stables, car ils s'appuient sur des contours contrastés.
- Les points de la **mâchoire** sont les moins précis : ce contour n'a pas d'arête nette et dépend du fond, des cheveux ou d'un voile.
- Les **5 points YuNet** (en bleu) doivent tomber au centre des yeux calculé par LBF. Dans le cas contraire, l'indicateur `lmk_consistency` augmente et signale un ajustement raté.
- Sur un visage tourné ou très incliné, le contour de la mâchoire « glisse » du côté caché, ce que mesure `yaw_index`.

### 7.2 Les descripteurs sont-ils bien normalisés ? (test d'invariance)
On vérifie la propriété recherchée. Pour une image de référence par sujet, on applique en mémoire, **sans rien écrire dans `data/`**, des transformations qui ne changent pas l'identité : rotation de 15°, réduction d'échelle, translation, éclairage plus sombre, léger flou. On compare ensuite les descripteurs :
- **coordonnées brutes en pixels** : elles devraient beaucoup varier ;
- **coordonnées normalisées** : elles devraient rester presque identiques.

La mesure est l'écart moyen par point, exprimé en fraction de l'IOD de l'image de référence. On l'appelle *NME*, *Normalised Mean Error*, la mesure classique en alignement facial.

```python
def transforms(img):
    h, w = img.shape[:2]
    yield "rotation 15°", cv2.warpAffine(img, cv2.getRotationMatrix2D((w / 2, h / 2), 15, 1.0), (w, h),
                                          borderMode=cv2.BORDER_REFLECT)
    yield "échelle x0.6", cv2.resize(img, None, fx=0.6, fy=0.6, interpolation=cv2.INTER_AREA)
    T = np.float32([[1, 0, 0.08 * w], [0, 1, 0.05 * h]])
    yield "translation", cv2.warpAffine(img, T, (w, h), borderMode=cv2.BORDER_REFLECT)
    yield "sombre (x0.5)", (img.astype(np.float32) * 0.5).astype(np.uint8)
    yield "flou gaussien", cv2.GaussianBlur(img, (0, 0), 1.5)

def extract_from_array(img):
    img, img_eq, _ = preprocess(img)
    f = detect_faces(img)
    f = f[f[:, 14] >= DET_SCORE_MIN] if len(f) else f
    if len(f) == 0: return None
    face = f[np.argmax(f[:, 2] * f[:, 3])]
    pts, _ = fit_landmarks(img_eq, face)
    if pts is None: return None
    pts_n, iod, _ = normalize_landmarks(pts)
    return pts, pts_n, iod

rows = []
# Référence = image de meilleure qualité de chaque sujet (pose la plus frontale, puis résolution la plus élevée)
refs = (okd.assign(_yaw=okd.yaw_index.abs(), _roll=okd.roll_deg.abs())
           .sort_values(["quality_ok", "_yaw", "_roll", "iod_px"], ascending=[False, True, True, False])
           .groupby("subject_id").head(1))
for _, r in refs.iterrows():
    base, _ = load_image(DATA_DIR / r.file_relpath)
    ref = extract_from_array(base)
    if ref is None: continue
    pts0, n0, iod0 = ref
    for name, timg in transforms(base):
        out = extract_from_array(timg)
        if out is None:
            rows.append({"sujet": r.subject_id, "transformation": name, "NME_brut": np.nan, "NME_normalisé": np.nan}); continue
        pts1, n1, _ = out
        rows.append({"sujet": r.subject_id, "transformation": name,
                     "NME_brut": np.linalg.norm(pts1 - pts0, axis=1).mean() / iod0,
                     "NME_normalisé": np.linalg.norm(n1 - n0, axis=1).mean()})
inv = pd.DataFrame(rows)
display(inv.round(4))
if len(inv):
    print(f"Écart moyen brut : {inv.NME_brut.mean():.3f} IOD | normalisé : {inv['NME_normalisé'].mean():.3f} IOD")
```

**Interprétation** : sur les coordonnées brutes, une rotation ou un changement d'échelle déplace les points de l'ordre de l'IOD ou davantage. Après normalisation, l'écart tombe à quelques centièmes d'IOD. Ce résidu ne vient pas de la normalisation, qui est exacte pour une similitude. Il vient du **bruit d'ajustement de LBF** (rééchantillonnage, flou, éclairage). Ce résidu est précisément la **variabilité intra-classe** dont parle le cours : la même personne ne produit jamais exactement le même gabarit.

## 8. Illustration : distances intra-sujet et inter-sujets
Ce n'est pas l'objet de la partie 1, mais c'est le lien avec la suite du module (vérification 1:1, score, seuil, FAR/FRR). On calcule une **distance euclidienne** entre gabarits (les 136 coordonnées normalisées) :
- **paires intra-sujet** (*genuine*), qui correspondent à la même personne ;
- **paires inter-sujets** (*impostor*), qui correspondent à des personnes différentes.

Plus les deux distributions sont séparées, plus le descripteur est **discriminant**, ce qui renvoie au critère d'unicité du cours.

> **Prudence ⚠️** : avec quelques images, les taux obtenus ne sont **pas** statistiquement significatifs au sens de l'ISO/IEC 19795-1. De plus, la géométrie de 68 points est **faiblement discriminante** comparée aux *embeddings* profonds (ArcFace, FaceNet) utilisés en reconnaissance faciale moderne. Seules les lignes `quality_ok` et non dupliquées sont utilisées.

```python
gal = okd[okd.quality_ok & ~okd.is_duplicate]
if gal.subject_id.nunique() < 2 or gal.groupby("subject_id").size().max() < 2:
    # Repli assumé et signalé : on exclut seulement les défauts graves (pose, ajustement raté, visages multiples)
    # et on tolère la basse résolution, pour illustrer le principe (sans valeur statistique).
    grave = okd.quality_issues.fillna("").str.contains("yaw|roll|landmark_fit|multiple_faces")
    gal = okd[~grave & ~okd.is_duplicate]
    print("Repli : pas assez de gabarits « quality_ok ». On utilise les images sans défaut grave "
          "(la basse résolution est tolérée).")
X = gal[LMK_COLS].to_numpy(); S = gal.subject_id.to_numpy()
gen, imp = [], []
for i in range(len(X)):
    for j in range(i + 1, len(X)):
        dist = float(np.linalg.norm(X[i] - X[j]) / np.sqrt(68))   # distance moyenne par point (en IOD)
        (gen if S[i] == S[j] else imp).append(dist)
print(f"{len(gal)} gabarits | {len(gen)} paires intra-sujet | {len(imp)} paires inter-sujets")

if gen and imp:
    gen, imp = np.array(gen), np.array(imp)
    thr = np.linspace(0, max(gen.max(), imp.max()) * 1.05, 400)
    FRR = np.array([(gen > t).mean() for t in thr])      # vrais utilisateurs rejetés
    FAR = np.array([(imp <= t).mean() for t in thr])     # imposteurs acceptés
    k = np.argmin(np.abs(FAR - FRR))
    fig, ax = plt.subplots(1, 2, figsize=(11, 3.6))
    bins = np.linspace(0, max(gen.max(), imp.max()) * 1.05, 16)   # mêmes classes pour les deux histogrammes
    ax[0].hist(gen, bins=bins, alpha=.7, label=f"intra-sujet (genuine, n={len(gen)})")
    ax[0].hist(imp, bins=bins, alpha=.7, label=f"inter-sujets (imposteur, n={len(imp)})")
    ax[0].set_xlabel("distance entre gabarits (IOD)"); ax[0].legend(); ax[0].set_title("Distributions de distances")
    ax[1].plot(thr, FAR, label="FAR"); ax[1].plot(thr, FRR, label="FRR")
    ax[1].axvline(thr[k], ls="--", c="gray"); ax[1].set_xlabel("seuil"); ax[1].legend()
    ax[1].set_title(f"EER ≈ {(FAR[k] + FRR[k]) / 2:.1%} (illustratif)")
    if min(len(gen), len(imp)) < 10:
        print(f"Attention : {len(gen)} paire(s) genuine et {len(imp)} paire(s) imposteur seulement. "
              "L'EER affiché n'a aucune valeur statistique (il faudrait des centaines de comparaisons ; cf. la « règle des 30 » de Doddington citée dans l'ISO/IEC 19795-1).")
    plt.tight_layout(); plt.show()
else:
    print("Il faut au moins 2 sujets avec 2 images de bonne qualité chacun pour cette illustration.")
```

## 9. Discussion : limites, sécurité et vie privée
### Limites techniques
- **Pose** : la normalisation ne corrige que le roulis. Un visage tourné (lacet) déforme le gabarit, d'où le contrôle `yaw_index`. Un alignement 3D (estimation de pose par `solvePnP` sur un modèle de tête) serait l'étape suivante.
- **Modèle LBF** : il a été entraîné sur 300-W, avec des visages majoritairement de face et des visages occidentaux sur-représentés. Il existe un **risque de biais** (âge, couleur de peau, port du voile ou de lunettes), limite citée dans le cours. Il faudrait mesurer le taux FTA par sous-groupe.
- **Expression** : bouche ouverte ou yeux plissés modifient `mouth_open_ratio` et l'EAR. C'est utile pour détecter la vivacité, mais cela ajoute de la variabilité intra-sujet.
- **LBF renvoie toujours 68 points**, même sur un mauvais visage : l'indicateur de cohérence YuNet/LBF est indispensable.
- **Aucun PAD** (*Presentation Attack Detection*, ISO/IEC 30107) : une photo de photo ou un écran produit des landmarks parfaitement valides. Ce pipeline ne doit **jamais** servir seul à un contrôle d'accès.

### Sécurité et protection du gabarit (ISO/IEC 24745)
`features_faces.csv` est un **fichier de gabarits biométriques**. Même sans image, la géométrie permet un rapprochement entre bases (*linkability*), et en partie une reconstruction. En exploitation réelle, il faudrait :
- **chiffrer le fichier au repos** (AES-256, clé gérée par une PKI ou un HSM) et en contrôler l'accès ;
- appliquer une **protection de gabarit** répondant aux trois exigences de l'ISO/IEC 24745 : irréversibilité, révocabilité et non-liaison (par exemple *cancelable biometrics*, chiffrement homomorphe) ;
- privilégier le **stockage local** (carte, terminal) et l'**authentification 1:1** plutôt que l'identification 1:N centralisée (CNDP 478-2013) ;
- fixer une **durée de conservation**, tenir le **registre des traitements** et documenter une **analyse d'impact (PIA)**, selon la gouvernance vue en cours.

# Conclusion
Le pipeline parcourt automatiquement les dossiers, traite chaque fichier sans s'interrompre en cas d'erreur, extrait 68 landmarks puis 136 coordonnées et 12 ratios **normalisés**, évalue la qualité, et exporte le tout dans `outputs/features_faces.csv`. Le test d'invariance confirme que la normalisation par similitude supprime l'effet de la position, de l'échelle et de la rotation dans le plan. Il reste une variabilité intra-sujet, due au bruit d'ajustement, que la phase de comparaison devra gérer par un **seuil**.
